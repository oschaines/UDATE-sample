```python
import pandas as pd
import math
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("UDatE_Teal.csv")
```

```python
meta = df.loc[:1,:]

country_cols = ['Q1_4_TEXT', 'Q1_5_TEXT', 'Q1_6_TEXT',
       'Q1_10_TEXT', 'Q1_7_TEXT', 'Q1_8_TEXT', 'Q1_9_TEXT']
df["Q1_TEXT"] = df[country_cols].bfill(axis=1).iloc[:, 0]

cols = ['Q1', 'Q1_TEXT', 'Q2', 'Q3', 'Q4',
       'Q4_6_TEXT', 'Q5', 'Q6', 'Z_register', 'Z_impact', 'EBE']

df = df.loc[2:, cols]
numeric_cols = ["Z_register", "Z_impact", "EBE"]
df[numeric_cols] = df[numeric_cols].astype(float)
df.head()
```

```python
df.loc[20:,"EBE"].hist()
```

## Analyse grouped by country:

```python
means = df[["Q1", "Z_register", "Z_impact", "EBE"]].groupby("Q1").mean()
counts = df[["Q1", "Z_register", "Z_impact", "EBE"]].groupby("Q1").count()
```

```python
fig, axs = plt.subplots(1,2)
axs[0].bar(counts.index, counts["EBE"])
axs[1].bar(means.index, means["EBE"])
plt.show()
```

## By gender:

```python
means = df[["Q2", "Z_register", "Z_impact", "EBE"]].groupby("Q2").mean()
counts = df[["Q2", "Z_register", "Z_impact", "EBE"]].groupby("Q2").count()
```

```python
fig, axs = plt.subplots(1,2)
axs[0].bar(counts.index, counts["EBE"])
axs[1].bar(means.index, means["EBE"])
plt.show()
```
